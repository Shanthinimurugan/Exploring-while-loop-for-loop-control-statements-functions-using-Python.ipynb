**Task 1:** While Loop & Control Statements -Number Guessing Game

**Problem Statement:** Create a Python program that implements a simple number guessing
game using a while loop. The program should make use of control statements such as else,
break, and continue.


```python
# method 1:(Using Random)

import random
secret2 = random.randint(1, 10)
attempts = 4
while attempts>0:
  Guess = int(input("Guess the number (between 1 and 10):"))
  if Guess < 1 or Guess > 10:
    print("Your guess is out of range.")
    continue
  if Guess < secret2:
    print("Too low")
  elif Guess > secret2:
    print("Too high")
  elif Guess == secret2:
    print("Congratulations! You guessed the correct number.")
    break
  attempts-=1
  print("Attempts remaining:", attempts)
else:
  print("Better luck next time!")
  print("The number is", secret2)
```

```python
# method 2: (without using random)

secret = 3
attempts = 4
while attempts>0:
  Guess = int(input("Guess the number (between 1 and 10):"))
  if Guess < 1 or Guess > 10:
    print("Your guess is out of range.")
    continue
  elif Guess < secret:
    print("Too low")
  elif Guess > secret:
    print("Too high")
  elif Guess == secret:
    print("Congratulations! You guessed the correct number.")
    break
  attempts-=1
  print("Attempts remaining:", attempts)
else:
  print("Better luck next time!")

```

**Task 2:** For Loop - Multiplication Table Generator

**Problem Statement:** Create a Python program that generates and prints a multiplication
table (from 1 to 10) for a given number using a for loop and the range function.

```python
a= int(input("Enter the number for which you want the multiplication table:"))

for j in range(1,11):
  print(f"{a}X{j}={a*j}")
```

**Task 3:** Function- BMI Calculator

**Problem Statement:** Create a Python program that calculates the Body Mass Index (BMI).

```python
def calculate_bmi(weight, height):
  BMI = weight/(height)**2
  return(f"{BMI:.2f}")
```

```python
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

bmi = calculate_bmi(weight,height)

print(f"Your BMI is: {bmi}")

# or print(f"Your BMI is: {bmi:.2f}")
```
